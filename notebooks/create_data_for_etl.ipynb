```python
import boto3
```

```python
session = boto3.Session(
    region_name='us-east-1',
    aws_access_key_id='test',
    aws_secret_access_key='test'
)

s3 = session.client("s3", endpoint_url="http://localhost:4566")
```

```python
s3.create_bucket(Bucket="data-platform-clean")
```

```python
raw_csv = """repo,pr_number,author_type
example/demo,1, HUMAN
example/demo,1,human
example/demo,2,Agent
example/demo,not-a-number,agent
example/demo,3,unknown
"""
```

```python
s3.put_object(
    Bucket="data-platform-raw",
    Key="etl/prs.csv",
    Body=raw_csv.encode("utf-8"),
)
```
