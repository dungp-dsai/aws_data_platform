```python
import boto3
```

```python
session = boto3.Session(
    region_name='us-east-1',
    aws_access_key_id='test',
    aws_secret_access_key='test'
)
```

```python
endpoint = 'http://localhost:4566'
```

```python
lambda_client = session.client('lambda', endpoint_url=endpoint)
```

```python
with open('function.zip', 'rb') as f:
    zip_bytes = f.read()
    
lambda_client.create_function(
    FunctionName='pr-events-ingestor',
    Runtime='python3.12',
    Role='arn:aws:iam::000000000000:role/lambda-kinesis-role',
    Handler='lambda_function.lambda_handler',
    Code={
        'ZipFile': zip_bytes,
    },
    Description='This is the lambda function to proceed items on stream',
    Environment={
        'Variables': {
            'FLOCI_ENDPOINT': 'http://host.docker.internal:4566',
            'RAW_BUCKET': 'data-platform-raw'
        }
    },
    Timeout=30
)
```
