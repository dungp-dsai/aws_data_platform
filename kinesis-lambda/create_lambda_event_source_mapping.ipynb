```python
import boto3
```

```python
session = boto3.Session(
    region_name = 'us-east-1',
    aws_access_key_id = 'test',
    aws_secret_access_key = 'test'
)

endpoint = 'http://localhost:4566'
```

```python
kinesis = session.client('kinesis', endpoint_url=endpoint)
```

```python
stream_arn = kinesis.describe_stream_summary(StreamName='pr-events-stream')['StreamDescriptionSummary']['StreamARN']
```

```python
stream_arn
```

```python
lambda_client = session.client('lambda', endpoint_url=endpoint)
```

```python
lambda_client.create_event_source_mapping(
    FunctionName='pr-events-ingestor',
    EventSourceArn=stream_arn,
    StartingPosition='LATEST',
    BatchSize=5
)
```
